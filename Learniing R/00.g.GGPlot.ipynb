# GGPlot2

```r
library(tidyverse)
```

```r
nhanes <-  read_csv("data/nhanes.csv")
```

```r
nhanes |> head()
```

```r
#ggplot creates a canvas on which to print a plot
nhanes |> ggplot()
```

```r
# Aesthetics abbreviates aes allows us to use our dataset to set parameters for our plot like x- axis, y-axis, etc
nhanes |> ggplot() + aes(x=Age)
```

```r
# there are several geometries that will enable us plot example geom_histogram, geom_boxplot etc
# geom_hoistogram lets you set number of bars in the bistogram using bins
nhanes |> ggplot() + aes(x=Age) + geom_histogram(bins=20, color='white', fill='royalblue') + theme_minimal()
```

```r
#boxplot
nhanes |> ggplot() + aes(y=Age) + geom_boxplot(fill='royalblue') + theme_minimal()
```

```r
# density plot
# for geom_density you can set a bandwidth using the bw argument = a value to smoothen the rough plot
nhanes |> ggplot() + aes(x=BPDiaAve) + geom_density(fill='royalblue', bw=3) + theme_minimal() 
```

```r
# barcharts
nhanes |> ggplot() + aes(x= Race1) + geom_bar(fill='royalblue') + theme_minimal()
```

```r
# often combine geom_point for scatterplot with geom_smooth for regression line
# remember to add the argument method=lm to get a straight regression line

nhanes |> 
  filter(Age == 77)  |> 
  ggplot() + 
  aes(x=Weight, y=Height) + 
  geom_point(size=3, color='royalblue') + 
  geom_smooth(method=lm, color='maroon') + theme_minimal()
```

```r
nhanes |> 
  ggplot() + 
  aes(x=Race1, y=Height, fill=Race1) + 
  geom_boxplot() + 
  theme_minimal() + 
  scale_fill_manual(values=c('green', 'turquoise', 'cyan', 'magenta', 'maroon'))
```

```r
nhanes |> 
  ggplot() + aes(x=Race1, fill=Education) + geom_bar(position='dodge', color='white') + theme_minimal()
```
