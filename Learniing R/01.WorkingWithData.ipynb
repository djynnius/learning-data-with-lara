# Working with Data

### Data formats 
- **Coma Separated Values** with extension .csv These are very universal and can be imported into almost any data management application, they can be imported for data analysis, or into a database and can even be opened by Excel or other spreadsheet applications like Google Sheets and others. They typically have values separated by coma and the first line is usueally the variable names and referred to as the header.
- Variants of CSV include **tab separated values** .tsv
- Recently a file format named **parquet** has become popular. Their extension is .parquet
- Common for web exchange - **Java Script Object Notation** .json files
- Proprietary data extension: 
    + Microsoft excel - .xls, .xlsx files
    + R - .RData, .RDS
    + SAS - .sas7bdat
    + Stata - .xport, .xpt

### How R reads these files

- CSV is read by default using **read.csv()**
    ```R
    df <- read.csv("/path/to/file.csv")
    ```
- TSV can be read with **read.table()**
- Tidyverse offers an alternative for importing csv called **read_csv()** 
    ```R
    library(tidyverse)

    df <- read_csv("/path/to/file.csv")
    ```
- To import parquet you need a package called arrow (install.packages("arrow"))
    ```R
    library(arrow)
    df <- read_parquet("/path/to/file.parquet")
    ```
- For most other data types you need some library to import them, one such is called foreign andother i haven and they allow for several types of files

- For R RData and RDS
    ```R
    df1 <- read_rds("/path/to/file.rds")
    load("/path/to/file.rdata")

    ```
