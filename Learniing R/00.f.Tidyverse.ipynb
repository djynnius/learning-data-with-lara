### Filter

````r
# Wranglin data with tidyverse
nhanes |> filter(Age > 18)
````

````r
nhanes |> filter(Age >= 13 & Age <= 19)
````

````r
nhanes |> filter(Age %in% c(13, 14, 15, 16, 17, 18, 19))
````

````r
13:19
````

````r
nhanes |> filter(Gender == 'male')
````

````r
nhanes |> filter(Race1 == 'White')
````

````r
# multiple filter using the & operator can be done with coma separation
nhanes |> filter(Race1 == 'Black' & Age < 18 & Gender == 'male')
````

````r
nhanes |> filter(Race1 == 'Black', Age < 18, Gender == 'male')
````

````r
nhanes |> filter(Race1 %in% c('White', 'Black'))
````

````r
# Selecting specific columns 
nhanes |> select(Age, Gender, Race1, Weight, Height)
````

````r
#Drop some columns 
nhanes |> select(-Age, -Gender, -Race1, -Education)
````

````r
# rename variables while selecting, new name = old name
nhanes |> select(Age, Sex=Gender, Race=Race1, Education, Weight, Height)
````

````r
#renaming only without selecting
nhanes |> rename(Sex=Gender, Race=Race1, Ethnicity=Race3)
````

````r
# randomly select from the dataframe
nhanes |> sample_n(10)
````

````r
# Chaining
nhanes |> 
  filter(Age < 18, Gender=='male', Race1=='Black') |> 
  select(Sex=Gender, Race=Race1, Weight, Height, Age) |> 
  sample_n(100)
````

````r
# sort using arrange
nhanes |> 
  filter(Age < 18, Gender=='male', Race1=='Black') |> 
  select(Sex=Gender, Race=Race1, Weight, Height, Age) |> 
  sample_n(100) |> 
  arrange(Weight |> desc())
````

````r
# Aggregation
# summarize
# group_by 


#mean, 
#median, 
#sd
#sum
#IQR, 
#min
#max
#quantile

#n_distinct --distinct values
#n -- frequency


nhanes |> 
  summarize(
    AvgAge=mean(Age), 
    StDevAge=sd(Age), 
    AvgWeight=mean(Weight, na.rm=T), 
    AvgHeight=mean(Height, na.rm=T), 
    TotalWeight=sum(Weight, na.rm=T)
)
````

````r
nhanes |> summarize(Observations=n(), Races=n_distinct(Race1), Sexes=n_distinct(Gender))
````

````r
nhanes |> 
  group_by(Race1) |> 
  summarize(
    Count=n(), 
    MinWeight=min(Weight, na.rm=T),
    Q1Weight=quantile(Weight, 0.25, na.rm=T), 
    MedianWeight=median(Weight, na.rm=T), 
    Q3Weight=quantile(Weight, 0.75, na.rm=T), 
    MaxWeight=max(Weight, na.rm=T)
)
````

````r
nhanes |> 
  group_by(Gender) |> 
  summarize(
    Count=n(), 
    MinWeight=min(Weight, na.rm=T),
    Q1Weight=quantile(Weight, 0.25, na.rm=T), 
    MedianWeight=median(Weight, na.rm=T), 
    Q3Weight=quantile(Weight, 0.75, na.rm=T), 
    MaxWeight=max(Weight, na.rm=T)
)
````

````r
nhanes |> 
  group_by(Gender, Race1) |> 
  summarize(
    Count=n(), 
    MinWeight=min(Weight, na.rm=T),
    Q1Weight=quantile(Weight, 0.25, na.rm=T), 
    MedianWeight=median(Weight, na.rm=T), 
    Q3Weight=quantile(Weight, 0.75, na.rm=T), 
    MaxWeight=max(Weight, na.rm=T)
)
````

````r
# Univariate Analysis

nhanes |> with(Age) |> hist()
````

````r
nhanes |> with(Age) |> summary()
````

````r
# Categorical

nhanes |> with(Race1) |> table()
````

````r
nhanes |> with(Race1) |> table() |> prop.table() |> round(2)
````

````r
# Bivariate 

#Num vs Num

nhanes |> select(Weight, Height) |> cor(use = 'complete.obs')
````

````r
nhanes |> 
  filter(Age >= 18) |> # restricting to only adults
  select(Weight, Height) |> # looking at weight and height
  cor(use = 'complete.obs') # 
````

````r
nhanes |> select(Weight, Height) |> plot()
````

````r
# Num vs Cat 

nhanes |> with(Height) |> hist()
````

````r
nhanes |> 
  group_by(Race1) |> 
  summarise(
    Center = median(Height, na.rm=T), 
    Spread = IQR(Height, na.rm=T)
  )
````
