##### Potential aggregation

Numeric
- Min
- Max 
- Quantile
- Average = mean
- Median 50th percentile or Q2
- sum

Categorical
- frequency - n
- proportion/percentage


```r
#aggregation works great with summarize or summerise 
iris |> summarize(
  Minimum=min(Petal.Length), 
  Maximum=max(Petal.Length), 
  Median=median(Petal.Length), 
  Q1=quantile(Petal.Length, 0.25), 
  Q3=quantile(Petal.Length, 0.75), 
  Average=mean(Petal.Length), 
  Standard.Deviation=sd(Petal.Length), 
  IQR=IQR(Petal.Length), 
  Total=sum(Petal.Length)
) |> tibble()
```

```r
iris |> summarise(
  Count=n()
)
```

```r
iris |> 
  group_by(Species) |> 
  summarise(
    Count=n(), 
    Average = mean(Petal.Length), 
    Total = sum(Petal.Length)
  )
```

```r
penguins |> 
  group_by(species, sex) |> 
  summarize(
    Count=n(), 
    Average=mean(body_mass, na.rm=T)
  ) -> penguins.aggregated
```

```r
penguins |> 
  group_by(species, island, sex) |> 
  summarize(
    Count=n(), 
    `Average body mass`=mean(body_mass, na.rm=T)
  ) 
```

### Transforming aggregated data

```r
penguins.aggregated
```

```r
#flipping data axis
penguins.aggregated |> t()
```

```r
#making data wide
penguins.aggregated |> pivot_wider(names_from=sex, values_from=c(Count, Average)) -> penguins.summary
penguins.summary
```

```r
#cleaning wide data
penguins.summary |> 
  select(species, female=Count_female, male=Count_male, female_bm=Average_female, male_bm=Average_female)
```

```r
#making wide data long
penguins.summary |> pivot_longer(cols=c(starts_with('Count'), starts_with('Average')), names_to='Summary', values_to='Values')
```
